```python
import cv2
import os

# Path to input video
video_path = "./video/mine.mp4"

# Output directory (same folder as video)
output_dir = os.path.dirname(video_path)

# Create output folder for frames (optional but cleaner)
frames_dir = os.path.join(output_dir, "frames")
os.makedirs(frames_dir, exist_ok=True)

# Open video
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    raise IOError("Error: Cannot open video file")

frame_count = 0

while True:
    success, frame = cap.read()
    if not success:
        break

    # Save frame as image
    frame_filename = os.path.join(frames_dir, f"frame_{frame_count:05d}.png")
    cv2.imwrite(frame_filename, frame)

    frame_count += 1

cap.release()

print(f"✅ Extracted {frame_count} frames into folder: {frames_dir}")
```

```python
# remove every second frame and reset the indexes on the remaining frames

import os
import shutil
from pathlib import Path

frames_dir = "./video/frames"

# Get all frame files sorted
frame_files = sorted([f for f in os.listdir(frames_dir) if f.startswith("frame_") and f.endswith(".png")])

print(f"Found {len(frame_files)} frames")

# Create a temporary directory to hold the frames we want to keep
temp_dir = "./video/frames_temp"
os.makedirs(temp_dir, exist_ok=True)

# Keep every other frame (indices 0, 2, 4, 6, ...)
kept_frames = []
for i, frame_file in enumerate(frame_files):
    if i % 2 == 0:  # Keep even-indexed frames
        kept_frames.append(frame_file)
        src = os.path.join(frames_dir, frame_file)
        dst = os.path.join(temp_dir, frame_file)
        shutil.copy2(src, dst)

print(f"Keeping {len(kept_frames)} frames (every other frame)")

# Delete all original frames
for frame_file in frame_files:
    os.remove(os.path.join(frames_dir, frame_file))

# Rename and move back the kept frames with sequential numbering
for new_idx, old_frame in enumerate(kept_frames):
    old_path = os.path.join(temp_dir, old_frame)
    new_name = f"frame_{new_idx:05d}.png"
    new_path = os.path.join(frames_dir, new_name)
    shutil.move(old_path, new_path)

# Remove the temporary directory
os.rmdir(temp_dir)

print(f"Done! Renamed {len(kept_frames)} frames from 0 to {len(kept_frames)-1}")

# Verify the result
final_frames = sorted([f for f in os.listdir(frames_dir) if f.startswith("frame_") and f.endswith(".png")])
print(f"Final frame count: {len(final_frames)}")
print(f"First frame: {final_frames[0]}")
print(f"Last frame: {final_frames[-1]}")
```

```python
# remove first 10 frames and reset the indexes

import os
import shutil

frames_dir = "./video/frames"

# Get all frame files sorted
frame_files = sorted([f for f in os.listdir(frames_dir) if f.startswith("frame_") and f.endswith(".png")])

print(f"Found {len(frame_files)} frames")
print(f"First frame: {frame_files[0]}")
print(f"Last frame: {frame_files[-1]}")

# Skip first 10 frames
frames_to_keep = frame_files[10:]

print(f"\nRemoving first 10 frames...")
print(f"Keeping {len(frames_to_keep)} frames")

# Create a temporary directory
temp_dir = "./video/frames_temp"
os.makedirs(temp_dir, exist_ok=True)

# Copy frames to keep to temp directory
for frame_file in frames_to_keep:
    src = os.path.join(frames_dir, frame_file)
    dst = os.path.join(temp_dir, frame_file)
    shutil.copy2(src, dst)

# Delete all original frames
for frame_file in frame_files:
    os.remove(os.path.join(frames_dir, frame_file))

# Rename and move back with sequential numbering starting from 0
for new_idx, old_frame in enumerate(frames_to_keep):
    old_path = os.path.join(temp_dir, old_frame)
    new_name = f"frame_{new_idx:05d}.png"
    new_path = os.path.join(frames_dir, new_name)
    shutil.move(old_path, new_path)

# Remove the temporary directory
os.rmdir(temp_dir)

print(f"\nDone! Renamed {len(frames_to_keep)} frames from 0 to {len(frames_to_keep)-1}")

# Verify the result
final_frames = sorted([f for f in os.listdir(frames_dir) if f.startswith("frame_") and f.endswith(".png")])
print(f"Final frame count: {len(final_frames)}")
print(f"First frame: {final_frames[0]}")
print(f"Last frame: {final_frames[-1]}")
```
