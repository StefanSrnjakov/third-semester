```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('primer1.jpg', cv2.IMREAD_GRAYSCALE)
crop_128 = img[100:228, 100:228]
crop_512 = img[100:612, 100:612]

kernel = np.ones((5,5), np.uint8)
```

```python
opening_128 = cv2.morphologyEx(crop_128, cv2.MORPH_OPEN, kernel)
closing_128 = cv2.morphologyEx(crop_128, cv2.MORPH_CLOSE, kernel)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(crop_128, cmap='gray')
axes[0].set_title('Original 128x128')
axes[0].axis('off')
axes[1].imshow(opening_128, cmap='gray')
axes[1].set_title('Opening')
axes[1].axis('off')
axes[2].imshow(closing_128, cmap='gray')
axes[2].set_title('Closing')
axes[2].axis('off')
plt.tight_layout()
plt.show()
```

```python
opening_512 = cv2.morphologyEx(crop_512, cv2.MORPH_OPEN, kernel)
closing_512 = cv2.morphologyEx(crop_512, cv2.MORPH_CLOSE, kernel)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(crop_512, cmap='gray')
axes[0].set_title('Original 512x512')
axes[0].axis('off')
axes[1].imshow(opening_512, cmap='gray')
axes[1].set_title('Opening')
axes[1].axis('off')
axes[2].imshow(closing_512, cmap='gray')
axes[2].set_title('Closing')
axes[2].axis('off')
plt.tight_layout()
plt.show()
```

```python
opening_full = cv2.morphologyEx(img, cv2.MORPH_OPEN, kernel)
closing_full = cv2.morphologyEx(img, cv2.MORPH_CLOSE, kernel)

fig, axes = plt.subplots(1, 3, figsize=(18, 6))
axes[0].imshow(img, cmap='gray')
axes[0].set_title('Original Full Image')
axes[0].axis('off')
axes[1].imshow(opening_full, cmap='gray')
axes[1].set_title('Opening')
axes[1].axis('off')
axes[2].imshow(closing_full, cmap='gray')
axes[2].set_title('Closing')
axes[2].axis('off')
plt.tight_layout()
plt.show()
```
